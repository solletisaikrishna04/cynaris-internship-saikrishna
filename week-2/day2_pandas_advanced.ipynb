```python
import pandas as pd
customers=pd.DataFrame({
    "customer_id":[1,2,3,4,5],
    "customer_name":["Alice","Bob","Charlie","David","Eva"],
})
print(customers)
```

```python
import pandas as pd
orders=pd.DataFrame({
    "customer_id":[1,2,1,3,4],
    "order_id":[101,102,103,104,105],
    "order_amount":[250,150,300,200,400],
})
print(orders)
```

```python
print(type(customers))
print(type(orders))
```

```python
result=pd.merge(customers,orders,on="customer_id",how="inner")
print(result)
```

```python
result_left=pd.merge(customers,orders,on="customer_id",how="left")
print(result_left)
```

```python
result_right=pd.merge(customers,orders,on="customer_id",how="right")
print(result_right)
```

```python
result_outer=pd.merge(customers,orders,on="customer_id",how="outer")
print(result_outer)
```

```python
import pandas as pd
sales_data=pd.DataFrame({
    "region":["North","South","North","South"],
    "products":["laptop","mobile","laptop","mobile"],
    "sales":[1000,1500,1200,1800],
})
print(sales_data)

```

```python
pivot=pd.pivot_table(sales_data, index="products", columns="region", values="sales", aggfunc="sum", fill_value=0)
print(pivot)
```

```python
import pandas as pd
wide_data=pd.DataFrame({
    "id":[1,2,3],
    "math_score":[85,90,78],
    "english_score":[92,88,95],
})
print(wide_data)
```

```python
long=pd.melt(wide_data, id_vars=["id"], value_vars=["math_score","english_score"], var_name="subject", value_name="score")
print(long)
```

```python
import pandas as pd
sales_data=pd.DataFrame({
    "region":["North","South","East","West","North","South"],
    "products":["laptop","laptop","mobile","mobile","tablet","tablet"],
    "sales":[1000,1500,1200,1800,800,900],
})
print(sales_data)
```

```python
print(sales_data.columns)
```

```python
result=sales_data.groupby("region").agg(total_sales=("sales","sum")).reset_index()
print(result)
```
