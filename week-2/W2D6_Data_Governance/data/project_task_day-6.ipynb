```python
# RULE 4: EMAIL MUST NOT BE EMPTY

# 1. Create the rule
expectation_email = gx.expectations.ExpectColumnValuesToNotBeNull(
    column="email"
)

# 2. Validate the dataset
result_email = batch.validate(expectation_email)

# 3. Display the result
print("Email validation result:")
print(result_email["success"])

# 4. Display the number of missing emails
print("Missing email count:")
print(result_email["result"]["unexpected_count"])
```

```python
# RULE 5: ANNUAL REVENUE MUST NOT BE NEGATIVE

# 1. Create the rule
expectation_revenue = gx.expectations.ExpectColumnValuesToBeBetween(
    column="annual_revenue",
    min_value=0
)

# 2. Validate the dataset
result_revenue = batch.validate(expectation_revenue)

# 3. Display the result
print("Revenue validation result:")
print(result_revenue["success"])

# 4. Display invalid count
print("Invalid revenue count:")
print(result_revenue["result"]["unexpected_count"])
```

```python
# RULE 6: STATE MUST BE VALID

# 1. Create the rule
expectation_state = gx.expectations.ExpectColumnValuesToBeInSet(
    column="state",
    value_set=[
        "Andhra Pradesh",
        "Telangana",
        "Karnataka",
        "Tamil Nadu",
        "Maharashtra"
    ]
)

# 2. Validate the dataset
result_state = batch.validate(expectation_state)

# 3. Display the result
print("State validation result:")
print(result_state["success"])

# 4. Display invalid count
print("Invalid state count:")
print(result_state["result"]["unexpected_count"])
```

```python
# RULE 7: BUSINESS NAME MUST NOT BE EMPTY

# 1. Create the rule
expectation_name = gx.expectations.ExpectColumnValuesToNotBeNull(
    column="business_name"
)

# 2. Validate the dataset
result_name = batch.validate(expectation_name)

# 3. Display the result
print("Business name validation:")
print(result_name["success"])

# 4. Display missing count
print("Missing business names:")
print(result_name["result"]["unexpected_count"])
```

```python
# RULE 8: PHONE NUMBER MUST BE VALID

# 1. Create the rule
expectation_phone = gx.expectations.ExpectColumnValuesToMatchRegex(
    column="phone",
    regex=r"^[6-9]\d{9}$"
)

# 2. Validate the dataset
result_phone = batch.validate(expectation_phone)

# 3. Display the result
print("Phone validation result:")
print(result_phone["success"])

# 4. Display invalid phone count
print("Invalid phone count:")
print(result_phone["result"]["unexpected_count"])
```

```python
import os

# Create reports folder
os.makedirs("../reports", exist_ok=True)

# Store validation results
results = {
    "Business ID not empty": result,
    "Age between 18 and 100": result_age,
    "Business ID unique": result_unique,
    "Email not empty": result_email,
    "Revenue not negative": result_revenue,
    "Valid state": result_state,
    "Business name not empty": result_name,
    "Phone format valid": result_phone
}

# Create HTML report
html = "<h1>Data Quality Validation Report</h1>"

for rule, validation in results.items():
    status = "PASSED" if validation["success"] else "FAILED"
    html += f"<p>{rule}: {status}</p>"

# Save report
with open("../reports/gx_validation_report.html", "w") as file:
    file.write(html)

print("HTML report created successfully!")
```

```python
# Count passed and failed checks

total_checks = len(results)

passed_checks = sum(
    validation["success"]
    for validation in results.values()
)

failed_checks = total_checks - passed_checks

# Display scorecard

print("DATA QUALITY SCORECARD")
print("----------------------")

print("Total checks:", total_checks)
print("Passed checks:", passed_checks)
print("Failed checks:", failed_checks)

# Calculate quality percentage

quality_score = (passed_checks / total_checks) * 100

print("Quality Score:", round(quality_score, 2), "%")
```

```python
recommendations = [
    "Correct invalid ages to be between 18 and 100.",
    "Investigate and remove duplicate business IDs.",
    "Collect missing email addresses."
]

print("DATA QUALITY IMPROVEMENT RECOMMENDATIONS")

for number, recommendation in enumerate(recommendations, start=1):
    print(number, recommendation)
```

```python
# Create a copy of our dataset
masked_df = df.copy()

# Mask owner names
masked_df["owner_name"] = masked_df["owner_name"].str[0] + "***"

# Mask email addresses
masked_df["email"] = masked_df["email"].str.replace(
    r"(?<=.).(?=.*@)", "*", regex=True
)

# Mask phone numbers
masked_df["phone"] = masked_df["phone"].astype(str).str[:-4] + "XXXX"

# Display masked data
print(masked_df[["owner_name", "email", "phone"]].head())
```

```python
masked_df.to_csv(
    "../reports/indian_business_masked.csv",
    index=False
)

print("Masked dataset saved successfully!")
```
