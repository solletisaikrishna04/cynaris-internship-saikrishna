```python
import pandas as pd
import numpy as np  
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.DataFrame({
    "Region": ["South", "North", "East", "West", "South",
               "North", "East", "West", "South", "North"],
    "Product": ["Laptop", "Mobile", "Laptop", "Mobile", "Mobile",
                "Laptop", "Mobile", "Laptop", "Laptop", "Mobile"],
    "Sales": [50000, 30000, 45000, 20000, 25000,
              60000, 15000, 40000, 55000, 35000],
    "Quantity": [5, 10, 4, 8, 12, 6, 5, 4, 7, 9],
    "Profit": [8000, 5000, 7000, 2000, 4000,
               10000, 1500, 6000, 9000, 5500]
})

print(data.head())
```

```python
numeric_data=data.select_dtypes(include="number")
correlation=numeric_data.corr()
sns.heatmap(correlation, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
print(correlation)
```

```python
data.describe()
```

```python
data.groupby("Region")["Sales"].sum()
```

```python
data.groupby("Region")["Sales"].sum().plot(kind="bar", color="skyblue")
plt.title("Total Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.show()
```

```python
product_sales = data.groupby("Product")["Sales"].sum()
product_sales
```

```python
data.groupby("Region")["Sales"].sum().sort_values(ascending=False)
```

```python
data.groupby("Product")["Profit"].sum().sort_values(ascending=False)
```

```python
data.groupby("Region")["Sales"].sum().sort_values(ascending=False)
```

```python
data.groupby("Product")["Quantity"].sum().sort_values(ascending=False)
```

```python
data.groupby("Region")["Sales"].mean().sort_values(ascending=False)
```

```python
## 5 Business Insights

1. **Sales by Product:** Laptop generates higher total sales than Mobile.

2. **Profit by Product:** Compare total profit to identify the more profitable product.

3. **Sales by Region:** Identify the region with the highest total sales.

4. **Quantity Sold:** Identify the product with the highest total quantity sold.

5. **Average Sales by Region:** Identify the region with the highest average sale per record.
```

```python
data.groupby("Product")["Profit"].sum().plot(kind="bar", color="lightgreen")
plt.title("Total Profit by Product")
plt.xlabel("Product")
plt.ylabel("Total Profit")
plt.show()
```

```python
data.groupby("Product")["Quantity"].sum().plot(kind="bar", color="orange")
plt.title("Total Quantity Sold by Product") 
plt.xlabel("Product")
plt.ylabel("Total Quantity Sold")
plt.show()
```

```python
data.groupby("Region")["Sales"].mean().plot(kind="bar", color="purple")
plt.title("Average Sales by Region")
plt.xlabel("Region")
plt.ylabel("Average Sales")
plt.show()
```
