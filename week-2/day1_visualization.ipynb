```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
print("Libraries imported successfully!")
```

```python
import matplotlib.pyplot as plt
months = ['January', 'February', 'March', 'April', 'May', 'June']
sales = [2500, 3000, 4000, 3500, 5000, 6000]
plt.plot(months, sales)
plt.title('Monthly Sales')
plt.xlabel('Months')
plt.ylabel('Sales')
plt.show()
```

```python
highest_sales=max(sales)
print("The highest sales value is:", highest_sales)
print (months[sales.index(highest_sales)])
```

```python
import matplotlib.pyplot as plt
categories=["electronics","clothing","home appliances","books"]
sales=[5000,3000,2000,1000]
colors = ['blue', 'green', 'red', 'orange']
plt.bar(categories,sales,color=colors)
plt.title("Sales by Category")
plt.xlabel("Categories")
plt.ylabel("Sales")
plt.bar_label(plt.gca().containers[0])
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.xticks(rotation=45)
plt.show()
```

```python
highest_sales = max(sales)
index=sales.index(highest_sales)
print("The highest sales value is:", highest_sales)
print("The category with the highest sales is:", categories[index])
```

```python
import matplotlib.pyplot as plt
categories=["electronics","clothing","home appliances"]
sales=[5000,3000,2000]
plt.barh(categories,sales)
plt.title("Sales by Category")
plt.xlabel("Sales")
plt.ylabel("Categories")
plt.show()
```

```python
import matplotlib.pyplot as plt
advertising=[1000, 1500, 2000, 2500, 3000]
sales=[10000, 12000, 15000, 18000, 20000]
plt.scatter(advertising, sales)
plt.title("Advertising vs Sales")
plt.xlabel("Advertising Spend")
plt.ylabel("Sales")
plt.show()
```

```python
import matplotlib.pyplot as plt
marks=[85, 90, 78, 92, 88]
plt.hist(marks, bins=5, edgecolor='black')
plt.title("Distribution of Marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show
```

```python
import matplotlib.pyplot as plt
categories=["electronics","clothing","home appliances","books"]
sales=[5000,3000,2000,1000]
plt.pie(sales, labels=categories, autopct='%1.1f%%',explode=(0.1, 0, 0, 0), shadow=True)
plt.title("Sales Distribution by Category")
plt.show()
```

```python
import matplotlib.pyplot as plt
marks=[85, 90, 78, 92, 88]
plt.boxplot(marks)
plt.title("Boxplot of Marks")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
months = ['January', 'February', 'March', 'April', 'May', 'June']
sales = [2500, 3000, 4000, 3500, 5000, 6000]
sns.set_theme(style="whitegrid")
plt.plot(months, sales, marker='o')
plt.title('Monthly Sales')
plt.xlabel('Months')
plt.ylabel('Sales')
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
sales=[2500, 3000, 4000, 3500, 5000, 6000]
sns.set_theme(style="whitegrid")
sns.histplot(sales)
plt.title('Sales Distribution')
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style="whitegrid")
months = ['January', 'February', 'March', 'April', 'May', 'June']
sales = [2500, 3000, 4000, 3500, 5000, 6000]
sns.lineplot(x=months, y=sales, marker='o')
plt.title('Monthly Sales')
plt.xlabel('Months')
plt.ylabel('Sales')
plt.annotate('Highest Sales', xy=('June', 6000))
plt.show
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style="whitegrid")
months = ['January', 'February', 'March', 'April', 'May', 'June']
sales = [2500, 3000, 4000, 3500, 5000, 6000]
plt.plot(months, sales, marker='o')
plt.title('Monthly Sales')
plt.xlabel('Months')
plt.ylabel('Sales')
plt.savefig('monthly_sales.png')
plt.show()
```
