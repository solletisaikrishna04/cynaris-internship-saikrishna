```python
marks = np.array([85, 90, 5, 92, 88, 95, 80, 75, 82])
print("mean:", np.mean(marks))
print("median:", np.median(marks))
print("standard deviation:", np.std(marks))
print("quartile 25:", np.percentile(marks, 25))
print("quartile 75:", np.percentile(marks, 75))
```

```python
import numpy as np
import time

# Step 1: Create Python lists
a = list(range(100000))
b = list(range(100000))

# Step 2: Start Python timer
start = time.perf_counter()

# Step 3: Add numbers using Python loop
python_result = []

for i in range(len(a)):
    python_result.append(a[i] + b[i])

# Step 4: Calculate Python execution time
python_time = time.perf_counter() - start

# Step 5: Convert Python lists into NumPy arrays
x = np.array(a)
y = np.array(b)

# Step 6: Start NumPy timer
start = time.perf_counter()

# Step 7: Add numbers using NumPy
numpy_result = x + y

# Step 8: Calculate NumPy execution time
numpy_time = time.perf_counter() - start

# Step 9: Display results
print("Python time:", python_time)
print("NumPy time:", numpy_time)
print("Speedup:", python_time / numpy_time)
print("First 5 results:", numpy_result[:5])
```
