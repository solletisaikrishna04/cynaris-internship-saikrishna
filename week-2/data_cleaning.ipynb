```python
sales=[100,200,300,400,500]
import pandas as pd
data_test=pd.Series(sales)
print(data_test)
```

```python
data_test.describe()
```

```python
sales=[1000,1200,1500,1800,2000,2500,50000]
data_test=pd.Series(sales)
```

```python
Q1 = data_test.quantile(0.25)
Q3 = data_test.quantile(0.75)
iqr = Q3 - Q1
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", iqr)
```

```python
lower_limit = Q1 - 1.5 * iqr
upper_limit = Q3 + 1.5 * iqr
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
```

```python
outliers = data_test[(data_test < lower_limit) | (data_test > upper_limit)]
print("Outliers:", outliers)
```

```python
print("number of outliers:", len(outliers))
```

```python
cleaned_data = data_test[(data_test >= lower_limit) & (data_test <= upper_limit)]
print("Cleaned Data:", cleaned_data)
```

```python
capped_data=data_test.clip(lower=lower_limit, upper=upper_limit)
print("Capped Data:", capped_data)
```
