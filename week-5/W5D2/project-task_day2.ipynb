```python
import pandas as pd

df = pd.read_csv(r"..\W5D1\indian_retail_sales.csv")

print(df.shape)
```

```python
date_columns = ['Date of Birth', 'Sales Date', 'Order Date', 'Ship Date']

for col in date_columns:
    df[col] = pd.to_datetime(df[col], errors='coerce')
```

```python
if not hasattr(pd.DataFrame, "applymap"):
    pd.DataFrame.applymap = lambda self, func: self.map(func)
```

```python
dfte = AV.AutoViz(
    filename='',
    sep=',',
    depVar='',
    dfte=df,
    header=0,
    verbose=1,
    lowess=False,
    chart_format='svg',
    max_rows_analyzed=150000,
    max_cols_analyzed=30
)
```

```python
print(dfte.shape)
print(dfte.head())
```
