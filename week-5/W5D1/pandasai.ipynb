```python
def ask_pandasai(question):
    try:
        result = sdf.chat(question)

        print("Result:", result)

        if isinstance(result, str) and (
            "Unfortunately" in result or "Traceback" in result
        ):
            logging.error(
                "PandasAI failed for question: %s\n%s",
                question,
                result
            )
            print("Error saved in log file.")
            return None

        return result

    except Exception as e:
        logging.exception("PandasAI query failed: %s", question)
        print("PandasAI Error:", e)
        return None
```
