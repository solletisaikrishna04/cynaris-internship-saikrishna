```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
data={
    "sales":[100,200,500,400,200,300,600,700,800,900],
    "city":["mumbai","delhi","bangalore","mumbai","delhi","bangalore","mumbai","delhi","bangalore","mumbai"]

}
df=pd.DataFrame(data)
```

```python
print("Mean:", df["sales"].mean())
print("Median:", df["sales"].median())
print("Mode:", df["sales"].mode())
print("Standard Deviation:", df["sales"].std())
print("Variance:", df["sales"].var())
```

```python
print(df["sales"].quantile([0.25, 0.5, 0.75]))
```

```python
plt.hist(df["sales"], bins=5, edgecolor='black')
plt.title("Sales distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

```python
sns.boxplot(x="city", y="sales", data=df)
plt.title("Sales comparison by city")
plt.show()
```
