```python
import numpy as np
import matplotlib.pyplot as plt

n=10
p=0.7

result=np.random.binomial(n,p)

plt.hist(result,bins=range(10),edgecolor='black')
plt.title("Binomial distribution")
plt.xlabel("Number of successes")
plt.ylabel("Frequency")
plt.show()
print("Number of successes:", result)
```

```python
import numpy as np
population=np.arange(1,101)
means=[]
for i in range(1000):
    sample=np.random.choice(population,size=20)
    average=np.mean(sample)
    means.append(average)
print("Mean of sample means:", np.mean(means))

plt.hist(means,bins=30,edgecolor='black')
plt.title("central limit theorem")
plt.xlabel("Sample means")
plt.ylabel("Frequency")
plt.show()
```

```python
import pandas as pd
data={
    "like": [20, 15],
    "dislike": [10, 25]
}
df=pd.DataFrame(data,index=["boys","girls"])
total=df.values.sum()
print("Total:", total)

P_boys=df.loc["boys"].sum()/total
print("Probability of boys:", P_boys*100,"%")
p_boys_like=df.like.sum()/total
print("Probability of boys liking:", p_boys_like*100,"%")
```

```python
import numpy as np
import matplotlib.pyplot as plt
data=np.random.normal(loc=50,scale=10,size=1000)
plt.hist(data,bins=30,edgecolor='black')
plt.title("Normal Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```
