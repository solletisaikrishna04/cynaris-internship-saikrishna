```python
import numpy as np
import scipy.stats as stats
```

```python
delivery_times=[4.6,2.1,3.5,4.2,3.9,5.1,4.8,3.7,4.0,4.3]
claimed_mean=4.0
mean=np.mean(delivery_times)
print("Average delivery time:", mean)
```

```python
delivery_times=[4.6,2.1,3.5,4.2,3.9,5.1,4.8,3.7,4.0,4.3]
claimed_mean=4.0
t_stat, p_value=stats.ttest_1samp(delivery_times, claimed_mean)
print(np.mean(delivery_times))
print(t_stat)
print(p_value)
if p_value<0.05:
    print("reject the null hypothesis")
else:
    print("fail to reject the null hypothesis")
```

```python
bengaluru=[4,3,5,4,3,4,5,4,3,4]
hyderabad=[3,4,4,5,4,3,4,5,4,3]
print(np.mean(bengaluru))
print(np.mean(hyderabad))
t_stat, p_value=stats.ttest_ind(bengaluru, hyderabad)
print(t_stat)
print(p_value)
if p_value<0.05:
    print("reject the null hypothesis")
else:
    print("fail to reject the null hypothesis")
```

```python
delivery_times=[4.6,2.1,3.5,4.2,3.9,5.1,4.8,3.7,4.0,4.3]
mean=np.mean(delivery_times)
confidence_interval=stats.t.interval(0.95, len(delivery_times)-1, loc=mean, scale=stats.sem(delivery_times))
print("95% Confidence Interval:", confidence_interval)
print("Average delivery time:", mean)
```
