```python
import numpy as np
import pandas as pd
data={
     "adv": [1000,2000,3000,4000,5000],
    "sales": [5000,10000,7000,8000,2000]
}
df=pd.DataFrame(data)
correlation=df["adv"].corr(df["sales"])
correlation_2=df["adv"].corr(df["sales"], method='spearman')
print("Correlation between advertising and sales:", correlation)
print("Spearman correlation between advertising and sales:", correlation_2)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.heatmap(df.corr(), annot=True) 
plt.title("Correlation Heatmap")
plt.show()
```

```python
import matplotlib.pyplot as plt
plt.scatter(df["adv"], df["sales"])
plt.title("Advertising vs Sales")
plt.xlabel("Advertising")
plt.ylabel("Sales")
plt.show()
```

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
x=df[["adv"]]
y=df["sales"]
model.fit(x,y)

```

```python
new_data=pd.DataFrame({"adv":[6000]})
predicted_sales=model.predict(new_data)
print("Predicted sales for advertising of 6000:", predicted_sales)
```

```python
from sklearn.metrics import r2_score
predictions=model.predict(x)
r2=r2_score(y,predictions)
print("R-squared value:", r2)
```

```python
from sklearn.metrics import mean_absolute_error
mae=mean_absolute_error(y,predictions)
print("Mean Absolute Error:", mae)
```

```python
from sklearn.metrics import mean_squared_error
mse=mean_squared_error(y,predictions)
rmse=np.sqrt(mse)
print("Root Mean Squared Error:", rmse)
```

```python
residuals=y-predictions
print("Residuals:", residuals)
```

```python
plt.scatter(predictions,residuals)
plt.xlabel("Predicted Sales")
plt.ylabel("Residuals")
plt.title("Residuals vs Predicted Sales")
plt.axhline(y=0)
plt.show()
```
